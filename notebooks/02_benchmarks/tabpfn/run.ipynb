# TabPFN Benchmarks

Run all project datasets from a single notebook.

This notebook saves:
- fitted model objects to `models/<dataset>/tabpfn.joblib`
- benchmark metrics to `results/<dataset>/tabpfn.json`
- classification plots and reports for classification datasets

If TabPFN weights are not cached yet, make sure `TABPFN_TOKEN` is available in the environment before running the benchmark cell.

````python
import os

import pandas as pd

from benchmark.models.tabpfn_model import DATASET_TASKS, run_tabpfn_benchmark

DATASET_IDS = list(DATASET_TASKS)
COMMON_KWARGS = {
    "device": "auto",
    "n_estimators": 8,
    "predict_batch_size": 5000,
    "ignore_pretraining_limits": False,
    "downsample": False,
    "max_train_rows": 1000,
    "max_test_rows": 300,
}

os.environ["TABPFN_TOKEN"] = ""

print("TABPFN_TOKEN available:", bool(os.getenv("TABPFN_TOKEN")))
print("Datasets:", DATASET_IDS)
COMMON_KWARGS
````

````python
records = []
for dataset_id in DATASET_IDS:
    print(f"Running {dataset_id}...")
    output = run_tabpfn_benchmark(dataset_id, **COMMON_KWARGS)
    records.append(output)

summary = pd.DataFrame(records)
summary
````

````python
metric_cols = [
    "dataset_id",
    "task",
    "rmse",
    "accuracy",
    "auc_roc",
    "fit_time_s",
    "infer_time_per_sample_s",
    "model_path",
    "results_path",
]

summary.reindex(
    columns=[col for col in metric_cols if col in summary.columns]
).sort_values(by="dataset_id")
````

You can also rerun a subset by changing `DATASET_IDS`, for example:

```python
DATASET_IDS = ["abalone", "sleep_health"]
```

To run full data later on a stronger machine, set:

```python
COMMON_KWARGS["downsample"] = False
COMMON_KWARGS["max_train_rows"] = None
COMMON_KWARGS["max_test_rows"] = None
COMMON_KWARGS["n_estimators"] = 8
COMMON_KWARGS["predict_batch_size"] = 8
```
